```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

# Создаём словарь (справочник) с названиями перцентилей
# Словарь - это как коробка с двумя отделениями: ключ и значение
percentile_names = {
    0.25: "25 (Q1)",      # 0.25 = 25%, Q1 - первый квартиль
    0.50: "50 (Q2 - медиана)",  # 0.50 = 50%, это медиана
    0.75: "75 (Q3)",      # 0.75 = 75%, Q3 - третий квартиль
    0.90: "90",           # 0.90 = 90%
    0.95: "95",           # 0.95 = 95%
    1.00: "100 (максимум)" # 1.00 = 100%
}

with open('task5_percentiles.txt', 'w', encoding='utf-8') as f:
    f.write("=== ПЕРЦЕНТИЛИ ПО ВСЕМ СТОЛБЦАМ ===\n\n")
    
    for column in df.columns:
        # Пропускаем столбец с полом (он текстовый)
        if column == 'gender':
            continue  # continue = "перейди к следующему столбцу, этот пропусти"
        
        f.write(f"\n{column}:\n")
        
        # Для каждого перцентиля из нашего словаря
        for p in [0.25, 0.50, 0.75, 0.90, 0.95, 1.00]:
            value = df[column].quantile(p)  # quantile - это "перцентиль"
            # .quantile(0.25) спрашивает: "какое значение 25-го перцентиля?"
            
            if pd.notna(value):
                # Получаем название из словаря: percentile_names[p]
                f.write(f"  Перцентиль {percentile_names[p]}: {value:.2f}\n")

print("Перцентили сохранены в task5_percentiles.txt")
```
