```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

with open('task2_mean.txt', 'w', encoding='utf-8') as f:
    f.write("=== СРЕДНИЕ ЗНАЧЕНИЯ (MEAN) ===\n")
    f.write("(только для числовых столбцов)\n\n")
    
    for column in df.columns:
        try:
            mean_value = df[column].mean()
            if pd.notna(mean_value):
                f.write(f"{column}: {mean_value:.2f}\n")
        except:
            pass

print("Средние значения сохранены в файл task2_mean.txt")
```
