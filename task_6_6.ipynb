```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

males = df[df['gender'] == 'Male']      # Все мальчики
females = df[df['gender'] == 'Female']  # Все девочки

with open('task6_iqr_gender.txt', 'w', encoding='utf-8') as f:
    f.write("=== МЕЖКВАРТИЛЬНЫЙ РАЗМАХ (IQR) ПО ДЛИНЕ ТЕЛА ===\n\n")
    
    q1_male = males['length_cm'].quantile(0.25)  # 25-й перцентиль у мальчиков
    q3_male = males['length_cm'].quantile(0.75)  # 75-й перцентиль у мальчиков
    iqr_male = q3_male - q1_male                 # Вычитаем
    
    f.write("Самцы (Male):\n")
    f.write(f"  Q1 (25-й перцентиль): {q1_male:.2f} см\n")
    f.write(f"  Q3 (75-й перцентиль): {q3_male:.2f} см\n")
    f.write(f"  IQR: {iqr_male:.2f} см\n\n")
    
    q1_female = females['length_cm'].quantile(0.25)
    q3_female = females['length_cm'].quantile(0.75)
    iqr_female = q3_female - q1_female
    
    f.write("Самки (Female):\n")
    f.write(f"  Q1 (25-й перцентиль): {q1_female:.2f} см\n")
    f.write(f"  Q3 (75-й перцентиль): {q3_female:.2f} см\n")
    f.write(f"  IQR: {iqr_female:.2f} см\n")

print("IQR по полу сохранён в task6_iqr_gender.txt")
```
