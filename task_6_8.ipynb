```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

males = df[df['gender'] == 'Male']
females = df[df['gender'] == 'Female']

with open('task8_tusk_cv.txt', 'w', encoding='utf-8') as f:
    f.write("=== КОЭФФИЦИЕНТ ВАРИАЦИИ ДЛИНЫ КЛЫКОВ ===\n\n")
    
    mean_male = males['tusk_length_cm'].mean()   # средняя длина клыков у мальчиков
    std_male = males['tusk_length_cm'].std()     # стандартное отклонение
    cv_male = (std_male / mean_male) * 100       # коэффициент вариации
    
    f.write("Самцы (Male):\n")
    f.write(f"  Средняя длина клыков: {mean_male:.2f} см\n")
    f.write(f"  Стандартное отклонение: {std_male:.2f} см\n")
    f.write(f"  Коэффициент вариации: {cv_male:.2f}%\n\n")
    
    mean_female = females['tusk_length_cm'].mean()
    std_female = females['tusk_length_cm'].std()
    cv_female = (std_female / mean_female) * 100
    
    f.write("Самки (Female):\n")
    f.write(f"  Средняя длина клыков: {mean_female:.2f} см\n")
    f.write(f"  Стандартное отклонение: {std_female:.2f} см\n")
    f.write(f"  Коэффициент вариации: {cv_female:.2f}%\n")

print("Коэффициент вариации по клыкам сохранён в task8_tusk_cv.txt")
```
