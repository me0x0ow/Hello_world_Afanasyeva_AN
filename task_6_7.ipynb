```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

with open('task7_variability.txt', 'w', encoding='utf-8') as f:
    f.write("=== ПОКАЗАТЕЛИ ВАРИАЦИИ (РАЗБРОСА) ===\n")
    f.write("Дисперсия | Стандартное отклонение | Коэффициент вариации (%)\n\n")
    
    for column in df.columns:
        if column == 'gender':
            continue
        
        try:
            # Считаем три величины
            mean_val = df[column].mean()        # среднее (нужно для коэффициента)
            var_val = df[column].var()          # дисперсия
            std_val = df[column].std()          # стандартное отклонение
            
            # Коэффициент вариации
            # Если среднее не равно нулю, считаем, иначе пишем "бесконечность"
            if mean_val != 0:
                cv_val = (std_val / mean_val) * 100
            else:
                cv_val = float('inf')  # бесконечность
            
            if pd.notna(var_val):
                f.write(f"{column}:\n")
                f.write(f"  Дисперсия: {var_val:.2f}\n")
                f.write(f"  Стандартное отклонение: {std_val:.2f}\n")
                f.write(f"  Коэффициент вариации: {cv_val:.2f}%\n\n")
        except:
            pass

print("Показатели вариации сохранены в task7_variability.txt")
```
