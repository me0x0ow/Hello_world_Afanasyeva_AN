```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

with open('task3_median.txt', 'w', encoding='utf-8') as f:
    f.write("=== МЕДИАННЫЕ ЗНАЧЕНИЯ (MEDIAN) ===\n")
    f.write("(только для числовых столбцов)\n\n")
    
    for column in df.columns:
        try:
            median_value = df[column].median()
            if pd.notna(median_value):
                f.write(f"{column}: {median_value:.2f}\n")
        except:
            pass

print("Медианные значения сохранены в task3_median.txt")
```
