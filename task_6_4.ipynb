```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

with open('task4_mode.txt', 'w', encoding='utf-8') as f:
    f.write("=== МОДАЛЬНЫЕ ЗНАЧЕНИЯ (MODE) ===\n")
    f.write("(для ВСЕХ столбцов, включая текстовые)\n")
    f.write("(если несколько мод - выводятся через запятую)\n\n")
    
    for column in df.columns:
        # Находим моду(ы)
        modes = df[column].mode()
        # modes - это может быть одно значение, а может быть несколько
        
        # СЧИТАЕМ, сколько мод получилось
        if len(modes) == 0:
            # len() - длина. Если длина = 0, значит нет ни одной моды
            modes_str = "Нет данных (все значения уникальны)"
        elif len(modes) == 1:
            # Если ровно одна мода
            modes_str = str(modes[0])  # превращаем в текст
        else:
            # Если несколько мод - склеиваем их через запятую
            # modes.values - все значения мод
            # for v in modes.values - для каждого значения v
            # str(v) - превращаем в текст
            # ", ".join(...) - склеиваем через запятую
            modes_str = ", ".join(str(v) for v in modes.values)
        
        # Пишем в файл
        f.write(f"{column}: {modes_str}\n")

print("Модальные значения сохранены в task4_mode.txt")
```
